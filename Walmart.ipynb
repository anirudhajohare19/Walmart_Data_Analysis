```python
# Importing the necessary libraries
import pandas as pd
import numpy as np

import pymysql # works as my sql adaptor
from sqlalchemy import create_engine
```

```python
df = pd.read_csv('Walmart.csv')

df.shape
```

### There are currently 10051 rows and 11 columns in the dataset before cleanig 

```python
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
# Checking Duplicates records
df.duplicated().sum()
```

### There are 51 duplicates in the dataset so we are Removing them

```python
df.drop_duplicates(inplace=True)
```

```python
# Checking for null values
df.isnull().sum()
```

### There are 31 Null values in both unit_price and quantity columns so we are removing the null records

```python
df.dropna(inplace=True)
```

```python
df.shape
```

###  After removing the null values there are 9969 rows and 11 columns in the dataset

```python
# checking the data types of the columns
df.dtypes
```

```python
# changing the data type of the unit_price columns to float and removing the $ sign

df['unit_price'] = df['unit_price'].str.replace("$", "").astype(float)
```

```python
df.head()
```

```python
df.columns
```

```python
# Coverting the column names to lower case
df. columns = df. columns.str.lower()
```

```python
# Creating new Feature to calculate the total sales
df['Total_sales'] = df['unit_price'] * df['quantity']
df.head()
```

### New Column Created to calculate the total sales (Unit_price * Quantity)

```python
df.to_csv('Walmart_cleaned.csv', index=False)
```
